**16) Any columns containing 't' and 'f' as values (True, False), should be converted to 1, 0.**

```python
tf_columns = []

for col in df.columns:
    unique_vals = df[col].dropna().unique()
    if set(unique_vals).issubset({'t', 'f'}):
        tf_columns.append(col)

tf_columns
```

```python
for col in tf_columns:
    df[col] = df[col].map({'t': 1, 'f': 0})
```
